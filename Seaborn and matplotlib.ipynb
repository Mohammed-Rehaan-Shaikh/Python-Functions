```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

bar chart vertical

```python
# Please use numpy for estimator
def bar_vertical(data, x, y, ax=None, figsize = None, title=None, hue = None, estimator = None, xlab = None, ylab = None, show = None, decimals = None):
    if ax is None:
        if figsize is not None:
            plt.figsize = figsize
        ax = plt.gca()

    if hue and estimator is None:
        sns.countplot(data=data, x=x, y = y, ax=ax)
    elif hue is not None and estimator is None:
        sns.countplot(data=data, x=x, y = y, hue=hue, ax=ax)
    elif hue is None and estimator is not None:
        sns.barplot(data = data, x = x, y = y, estimator = estimator, ax = ax)
    else:
        sns.barplot(data = data, x = x, y = y, hue = hue, estimator = estimator, ax = ax)

    if decimals is None:
        if estimator is None:
            for i in range(len(data)):
                plt.text(i, data[y][i] + 0.5, f"{data[y][i]}", ha = 'center', va = 'bottom')
        else:
            for i, z in enumerate(data[x].unique()):
                value = estimator(data[y][data[x] == z])
                plt.text(i, value + 0.5, f"{value:}", ha = 'center', va = 'bottom')
    else:
        if estimator is None:
            for i in range(len(data)):
                plt.text(i, data[y][i] + 0.5, f"{round(data[y][i], decimals)}", ha = 'center', va = 'bottom')
        else:
            for i, z in enumerate(data[x].unique()):
                value = estimator(data[y][data[x] == z])
                plt.text(i, value + 0.5, f"{round(value, decimals)}", ha = 'center', va = 'bottom')
    
    if title is not None:
        ax.set_title(title)
    
    if xlab is not None:
        ax.set_xlabel(xlab)
    
    if ylab is not None:
        ax.set_ylabel(ylab)
    
    ax.set_ylim(0, max(data[y]) + 7)
    if show is not None and show.lower() == 'yes':
        plt.show()
```

bar chart horizontal

```python
def bar_horizontal(data, x, y, ax=None, figsize= None, title=None, hue = None, xlab = None, ylab = None, show = None, xlim = None):
    if ax is None:
            if figsize is not None:
                plt.figsize = figsize
            ax = plt.gca()

    if hue is not None:
        sns.countplot(data=data, x=x, y=y, hue=hue, ax=ax)
    else:
        sns.countplot(data=data, x=x, y=y, ax=ax)
    
    for i in range(len(data)):
        plt.text(data[y][i] + 0.5, i, f"{data[y][i]}", ha = 'center', va = 'bottom')
    
    if title is not None:
        ax.set_title(title)
    
    if xlab is not None:
        ax.set_xlabel(xlab)
    
    if ylab is not None:
        ax.set_ylabel(ylab)
    if xlim is not None:
        ax.set_xlim(xlim)
    if show is not None and show.lower() == 'yes':
        plt.show()
```

Count Chart Vertical

```python
def countplot_vertical(data, x, show = None, title=None, xlab=None, ylab=None,hue=None, figsize=None, ax=None):
    if ax is None:
        if figsize is not None:
            plt.figure(figsize=figsize)
        ax = plt.gca()
    if hue is not None:
        sns.countplot(data=data, x=x, hue=hue, ax=ax)
    else:
        sns.countplot(data=data, x=x, ax=ax)

    for i, value in enumerate(data[x].unique()):
        value_count = sum(data[x] == value)
        ax.text(
            i, value_count + 0.5,
            f"{value_count}",
            ha='center',
            va='bottom'
        )

    if title is not None:
        ax.set_title(title)

    if xlab is not None:
        ax.set_xlabel(xlab)

    if ylab is not None:
        ax.set_ylabel(ylab)

    ax.set_ylim(0, max(data[x].value_counts()) + 7)
    if show is not None and show.lower() == 'yes':
        plt.show()
```

Count Chart Horizontal

```python
def countplot_horizontal(data, y, show = None, ax=None, title=None,xlab=None, ylab=None, hue=None, figsize=None, xlim = None):

    if ax is None:
        if figsize is not None:
            plt.figure(figsize=figsize)
        ax = plt.gca()

    sns.countplot(
        data=data,
        y=y,
        hue=hue,
        ax=ax
    )

    for i, value in enumerate(data[y].unique()):
        value_count = (data[y] == value).sum()

        ax.text(
            value_count + 0.5,
            i,
            str(value_count),
            va='center'
        )

    if title is not None:
        ax.set_title(title)

    if xlab is not None:
        ax.set_xlabel(xlab)

    if ylab is not None:
        ax.set_ylabel(ylab)
    if xlim is not None:
        ax.set_xlim(xlim)
    if show is not None and show.lower() == 'yes':
        plt.show()
```

Line Chart With Labels

```python
def line_chart(data, x, y, ax=None, figsize = None, xlab = None, ylab = None, grid=None, title=None, marker = None, xlim = None, ylim = None, hue=None,
               left_up=None, left_center=None, left_down=None,
               right_up=None, right_center=None, right_down=None,
               center_up=None, center_center=None, center_down=None):
    if ax is not None:
        if figsize is not None:
            plt.figure(figsize=figsize)
        ax = plt.gca()
    if hue and marker is None:
        sns.lineplot(data=data, x=x,y=y,ax=ax)
    elif hue is not None and marker is None:
        sns.lineplot(data = data, x = x, y = y, hue = hue, ax =ax)
    elif hue is None and marker is not None:
        sns.lineplot(data = data, x = x, y = y, marker = marker, ax = ax)
    else:
        sns.lineplot(data = data, x = x, y = y, hue = hue, marker = marker, ax = ax)
    if grid is not None:
        plt.grid(grid)
    plt.gca().set_axisbelow(True)
```

Line Chart without Labels

```python
def line_chart(data, x, y, ax=None, fig_a=None, fig_b=None, grid=None, title=None, marker = None,):

    standalone = ax is None
    if standalone:
        if fig_a is None or fig_b is None:
            fig, ax = plt.subplots()
        else:
            fig, ax = plt.subplots(figsize=(fig_a, fig_b))

    if marker is None:
        sns.lineplot(data=data, x=x, y=y, ax=ax)
    else:
        sns.lineplot(data=data, x=x, y=y, ax=ax, marker = marker)

    ax.set_axisbelow(True)
    if grid is None:
        pass
    else:
        ax.grid(alpha=grid)
    if title is None:
        pass
    else:
        ax.set_title(title)

    if standalone:
        plt.show()
```

Subplots

```python
def subplots(plots, nrows=1, ncols=1, figsize=(10, 5), title=None):

    fig, axes = plt.subplots(
        nrows=nrows,
        ncols=ncols,
        figsize=figsize,
        squeeze=False
    )

    if title is not None:
        fig.suptitle(title)

    for i in range(len(plots)):
        plots[i](ax=axes[i // ncols, i % ncols])

    plt.tight_layout()
```
