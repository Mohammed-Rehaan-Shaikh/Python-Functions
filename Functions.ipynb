```python
import seaborn as sns
import matplotlib.pyplot as plt
```

bar chart vertical

```python
def bar_vertical(data, x, y, ax=None, fig_a=None, fig_b=None, grid=None, title=None):
    standalone = ax is None
    if standalone:
        if fig_a is None or fig_b is None:
            fig, ax = plt.subplots()
        else:
            fig, ax = plt.subplots(figsize=(fig_a, fig_b))

    sns.barplot(data=data, x=x, y=y, ax=ax)

    for i in range(len(data)):
        ax.text(
            i,
            data[y].iloc[i],
            f"{data[y].iloc[i]}",
            ha="center",
            va="bottom"
        )
    ax.set_axisbelow(True)
    if grid is None:
        pass
    else:
        ax.grid(alpha=grid)
    if title is None:
        pass
    else:
        ax.set_title(title)

    if standalone:
        plt.show()
```

bar chart horizontal

```python
def bar_horizontal(data, x, y, ax=None, fig_a=None, fig_b=None, grid=None, title=None):
    standalone = ax is None
    if standalone:
        if fig_a is None or fig_b is None:
            fig, ax = plt.subplots()
        else:
            fig, ax = plt.subplots(figsize=(fig_a, fig_b))

    sns.barplot(data=data, x=x, y=y, ax=ax)

    for i in range(len(data)):
        ax.text(
            data[x].iloc[i],
            i,
            f" {data[x].iloc[i]}",
            ha="left",
            va="center"
        )
    ax.set_axisbelow(True)
    if grid is None:
        pass
    else:
        ax.grid(alpha=grid)
    if title is None:
        pass
    else:
        ax.set_title(title)

    if standalone:
        plt.show()
```

Line Chart with Marker 'o'

```python
def line_chart(data, x, y, ax=None, fig_a=None, fig_b=None, grid=None, title=None,
               left_up=None, left_center=None, left_down=None,
               right_up=None, right_center=None, right_down=None,
               center_up=None, center_center=None, center_down=None):

    standalone = ax is None
    if standalone:
        if fig_a is None or fig_b is None:
            fig, ax = plt.subplots()
        else:
            fig, ax = plt.subplots(figsize=(fig_a, fig_b))

    sns.lineplot(data=data, x=x, y=y, ax=ax, marker = "o")

    # default empty lists so "in" checks never fail on None
    left_up = left_up or []
    left_center = left_center or []
    left_down = left_down or []
    right_up = right_up or []
    right_center = right_center or []
    right_down = right_down or []
    center_up = center_up or []
    center_center = center_center or []
    center_down = center_down or []

    for i in range(len(data)):
        x_val = data[x].iloc[i]
        y_val = data[y].iloc[i]

        if i in left_up:
            ax.text(x_val, y_val, f"{y_val}", ha="right", va="bottom")
        elif i in left_center:
            ax.text(x_val, y_val, f"{y_val}", ha="right", va="center")
        elif i in left_down:
            ax.text(x_val, y_val, f"{y_val}", ha="right", va="top")
        elif i in right_up:
            ax.text(x_val, y_val, f"{y_val}", ha="left", va="bottom")
        elif i in right_center:
            ax.text(x_val, y_val, f"{y_val}", ha="left", va="center")
        elif i in right_down:
            ax.text(x_val, y_val, f"{y_val}", ha="left", va="top")
        elif i in center_up:
            ax.text(x_val, y_val, f"{y_val}", ha="center", va="bottom")
        elif i in center_center:
            ax.text(x_val, y_val, f"{y_val}", ha="center", va="center")
        elif i in center_down:
            ax.text(x_val, y_val - 0.07, f"{y_val}", ha="center", va="top")
        else:
            ax.text(x_val, y_val + 0.06, f"{y_val}", ha="center", va="bottom")

    ax.set_axisbelow(True)
    if grid is None:
        pass
    else:
        ax.grid(alpha=grid)
    if title is None:
        pass
    else:
        ax.set_title(title)

    if standalone:
        plt.show()
```

Subplots

```python
def subplots(nrows, ncols, fig_a=None, fig_b=None, grid=None, title=None, plot=None):

    if fig_a is None or fig_b is None:
        fig, ax = plt.subplots(nrows=nrows, ncols=ncols)
    else:
        fig, ax = plt.subplots(
            nrows=nrows,
            ncols=ncols,
            figsize=(fig_a, fig_b)
        )

    if nrows == 1 and ncols == 1:
        ax_list = [ax]
    else:
        ax_list = ax.flat

    if plot is None:
        pass
    else:
        for i in range(len(plot)):
            plot[i](ax=ax_list[i])

    if title is None:
        pass
    else:
        fig.suptitle(title)

    plt.tight_layout()
    plt.show()
```
